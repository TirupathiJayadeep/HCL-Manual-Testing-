```python
'''1.Print all prime numbers between input range'''
start=int(input())
end=int(input())
lst=[]
for i in range(start,end+1):
    if i<2:
        continue
    for j in range(2,int(i**0.5)+1):
        if i%j==0:
            break
    else:
        lst.append(i)
print(lst)
```

```python
'''2.Factorial using recursion'''
def factorial(n):
    if n==0 or n==1:
        return 1
    else:
        return n*factorial(n-1)
n=int(input())
print(factorial(n))
```

```python
'''3.Square of numbers using lambda'''
lst=list(map(int,input().split(',')))
res=list(map(lambda x:x*x,lst))
print(res)
```

```python
'''4.Find the second largest element in a list'''
nums=list(map(int,input().split(',')))
largest=lst[0]
second_largest=float('-inf')
for i in nums:
    if i>largest:
        second_largest=largest
        largest=i
print(second_largest)
```

```python
'''5.Count frequency of characters in a string'''
s=input()
freq={}
for i in s:
    freq[i]=freq.get(i,0)+1
print(freq)
```

```python
'''6.Calculate area of a circle using math library.'''
import math
r=float(input())
area=math.pi*r*r
print(area)
```

```python
'''7.Reverse a string without using built‑in reverse'''
s=input()
res=""
for i in range(len(s)-1,-1,-1):
    res+=s[i]
print(res)
```

```python
'''8.Remove duplicates from a list'''
nums=input().split(',')
seen=set()
lst=[]
for i in nums:
    if i not in seen:
        seen.add(i)
        lst.append(i)
print(lst)
```

```python
'''9.Merge two dictionaries'''
d1={'name':{'Jai','Naveen'},'dept':{'AIML','CSE'}}
d2={'name':'Monish','dept':'AIML'}
for key in d2:
    d1[key].add(d2[key])
print(d1)
```

```python
'''10.Fibonacci series using recursion'''
def fibonacci(n):
    if n==0:
        return 0
    elif n==1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
n=int(input())
for i in range(n):
    print(fibonacci(i),end=" ")
    
```
