```python
'''12. Mobile Battery Management
A mobile application records the battery consumption and charging 
values during different time intervals. The system needs to identify 
the longest period during which the battery-related values continuously 
increase according to the recorded data.'''

lst=list(map(int,input().split(',')))
n=len(lst)
maximum=0
count=1
for i in range(1,n):
    if lst[i]>lst[i-1]:
        count+=1
    else:
        count=1
    maximum=max(maximum,count)     
print(maximum)
```

```python
'''17. Password Security Analysis
A security application receives a password and needs to identify
the longest portion of the password that reads the same from both directions.
This information is used as part of its password-pattern analysis.'''

s=input()
longest=1
long=1
n=len(s)
for i in range(n):
  j=n-1
  long=1
  while j>=i:
    if s[i:j+1]==s[i:j+1][::-1]:
      long=j-i+1
      longest=max(long,longest)
    j-=1
print(longest)
```

```python
'''14. Financial Investment Planning
An investor has several investment options, each with 
a required investment amount and an expected return. 
The investor has a fixed budget and can select each option 
at most once.Determine the maximum possible return that can
be obtained within the available budget.'''

investment=list(map(int,input().split(',')))
returns=list(map(int,input().split(',')))
budget=int(input())

dictn=dict(zip(investment,returns))
n=len(returns)
maximum=0
for i in range(n):
    for j in range(i+1,n):
        total=investment[i]+investment[j]
        if total<=budget:
            amount=dictn[investment[i]]+dictn[investment[j]]
            maximum=max(maximum,amount)
print(maximum)
            
```

```python
'''16. Island Mapping System
A geographical mapping application represents land
and water using a two-dimensional grid. Cells containing land
that are connected horizontally or vertically belong to the
same geographical region. The application needs to determine the
number of separate land regions present in the map.'''

grid=[]
n=int(input())
rows=input().split()
for row in rows:
    grid.append(list(map(int,row)))

rows=len(grid)
cols=len(grid[0])
count=0

def dfs(i,j):
    if i<0 or i>=rows or j<0 or j>=cols:
        return
    if grid[i][j]==0:
        return
    grid[i][j]=0
    dfs(i+1,j)
    dfs(i-1,j)
    dfs(i,j+1)
    dfs(i,j-1)
for i in range(rows):
    for j in range(cols):
        if grid[i][j]==1:
            count+=1
            dfs(i,j)
print(count)
```
